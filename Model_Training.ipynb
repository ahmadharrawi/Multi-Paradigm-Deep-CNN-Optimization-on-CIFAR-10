# ***-Data Ingestion & Preprocessing Cells***

Loading the raw CIFAR-10 data fields, standardizing the feature matrix elements to a continuous floating-point interval, and translating discrete labels into 10-dimensional categorical vectors.

```python
import keras
import matplotlib.pyplot as plt
from keras.models import Sequential, Model
from keras.layers import Input, Conv2D, MaxPooling2D, Flatten, Dense, Dropout, BatchNormalization
from keras.utils import to_categorical
from keras.datasets import cifar10

(X_train, y_train), (X_test, y_test) = cifar10.load_data()

print("Training data shape:", X_train.shape)
print("Testing data shape:", X_test.shape)
```

```python
X_train = X_train.astype('float32') / 255.0
X_test = X_test.astype('float32') / 255.0

y_train = to_categorical(y_train, 10)
y_test = to_categorical(y_test, 10)

num_classes = 10
print("Labels correctly preprocessed shape:", y_train.shape)
```

# ***-Spatial Data Augmentation Block***

Configuring real-time spatial transformations within the operational pipeline to build robust feature extraction layers against physical distortions.

```python
from keras.layers import RandomFlip, RandomRotation

def get_augmentation_layer():
    return keras.Sequential([
        RandomFlip("horizontal"),
        RandomRotation(0.05)
    ])
```

# ***-Model 1: Optimized Deep Custom CNN***

A specialized deep sequential network topology consisting of stacked convolutional blocks, batch norm layers, and spatial regularizers designed to isolate localized high-level abstractions.

```python
from keras.optimizers import Adam

def build_optimized_custom_cnn():
    model = Sequential()
    model.add(Input(shape=(32, 32, 3)))
    model.add(get_augmentation_layer())

    model.add(Conv2D(32, (3, 3), activation='relu', padding='same'))
    model.add(BatchNormalization())
    model.add(Conv2D(32, (3, 3), activation='relu', padding='same'))
    model.add(BatchNormalization())
    model.add(MaxPooling2D(pool_size=(2, 2)))
    model.add(Dropout(0.2))

    model.add(Conv2D(64, (3, 3), activation='relu', padding='same'))
    model.add(BatchNormalization())
    model.add(Conv2D(64, (3, 3), activation='relu', padding='same'))
    model.add(BatchNormalization())
    model.add(MaxPooling2D(pool_size=(2, 2)))
    model.add(Dropout(0.3))

    model.add(Conv2D(128, (3, 3), activation='relu', padding='same'))
    model.add(BatchNormalization())
    model.add(Conv2D(128, (3, 3), activation='relu', padding='same'))
    model.add(BatchNormalization())
    model.add(MaxPooling2D(pool_size=(2, 2)))
    model.add(Dropout(0.4))

    model.add(Flatten())
    model.add(Dense(256, activation='relu'))
    model.add(BatchNormalization())
    model.add(Dropout(0.5))
    model.add(Dense(num_classes, activation='softmax'))

    opt = Adam(learning_rate=0.0005)
    model.compile(optimizer=opt, loss='categorical_crossentropy', metrics=['accuracy'])
    return model

custom_cnn = build_optimized_custom_cnn()
custom_cnn.summary()
```

```python
print("Training the Optimized Custom CNN...")
history_custom = custom_cnn.fit(X_train, y_train, validation_data=(X_test, y_test), epochs=40, batch_size=128, verbose=1)

scores_custom = custom_cnn.evaluate(X_test, y_test, verbose=0)
print("\n>>> Final Stabilized CNN Test Accuracy: {:.2f}%".format(scores_custom[1] * 100))
```

# ***-Model 2: The Deep VGG-Style Architecture***

An advanced, multi-stage deep network mimicking a classic VGG configuration with paired convolutional groupings ($64 \rightarrow 128 \rightarrow 256$ feature maps). This model scales up capacity to cleanly interpret highly distorted augmented inputs without bottlenecking.

```python
import keras
from keras.models import Sequential
from keras.layers import Conv2D, MaxPooling2D, Flatten, Dense, Dropout, BatchNormalization, RandomFlip, RandomRotation, RandomZoom
from keras.optimizers import Adam

def build_90_plus_deep_vgg():
    model = Sequential()

    model.add(RandomFlip("horizontal", input_shape=(32, 32, 3)))
    model.add(RandomRotation(0.15))
    model.add(RandomZoom(0.1))

    model.add(Conv2D(64, (3, 3), activation='relu', padding='same'))
    model.add(BatchNormalization())
    model.add(Conv2D(64, (3, 3), activation='relu', padding='same'))
    model.add(BatchNormalization())
    model.add(MaxPooling2D((2, 2)))
    model.add(Dropout(0.25))

    model.add(Conv2D(128, (3, 3), activation='relu', padding='same'))
    model.add(BatchNormalization())
    model.add(Conv2D(128, (3, 3), activation='relu', padding='same'))
    model.add(BatchNormalization())
    model.add(MaxPooling2D((2, 2)))
    model.add(Dropout(0.35))

    model.add(Conv2D(256, (3, 3), activation='relu', padding='same'))
    model.add(BatchNormalization())
    model.add(Conv2D(256, (3, 3), activation='relu', padding='same'))
    model.add(BatchNormalization())
    model.add(MaxPooling2D((2, 2)))
    model.add(Dropout(0.45))

    model.add(Flatten())
    model.add(Dense(512, activation='relu'))
    model.add(BatchNormalization())
    model.add(Dropout(0.5))

    model.add(Dense(10, activation='softmax'))

    model.compile(optimizer=Adam(learning_rate=0.001),
                  loss='categorical_crossentropy',
                  metrics=['accuracy'])
    return model

final_deep_model = build_90_plus_deep_vgg()
final_deep_model.summary()
```

# ***-Model 2 Training & Execution Loop***

Generating comparative visualization plots mapping training and validation tracking metrics across epochs. These indicators allow us to verify the absolute suppression of overfitting barriers.

```python
from keras.callbacks import ReduceLROnPlateau

lr_scheduler = ReduceLROnPlateau(monitor='val_loss',
                                 factor=0.5,
                                 patience=3,
                                 min_lr=0.00001,
                                 verbose=1)

print("Training Deep VGG-Style Architecture for 90%+ Bonus Tier...")
history_deep = final_deep_model.fit(X_train, y_train,
                                    validation_data=(X_test, y_test),
                                    epochs=45,
                                    batch_size=64,
                                    callbacks=[lr_scheduler],
                                    verbose=1)

scores_deep = final_deep_model.evaluate(X_test, y_test, verbose=0)
print("\n>>> Deep VGG Final Test Accuracy: {:.2f}%".format(scores_deep[1] * 100))
```

# ***-Matplotlib Plotting***

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(16, 5))

plt.subplot(1, 2, 1)
plt.plot(history_deep.history['loss'], label='Training Loss', linewidth=2)
plt.plot(history_deep.history['val_loss'], label='Validation Loss', linestyle='--', linewidth=2)
plt.title('Model Loss Convergence Trends', fontsize=14, fontweight='bold')
plt.xlabel('Epochs Run', fontsize=11)
plt.ylabel('Loss Intensity Value', fontsize=11)
plt.grid(True, linestyle=':', alpha=0.6)
plt.legend(fontsize=11)

plt.subplot(1, 2, 2)
plt.plot(history_deep.history['accuracy'], label='Training Accuracy', color='green', linewidth=2)
plt.plot(history_deep.history['val_accuracy'], label='Validation Accuracy', color='tab:red', linestyle='--', linewidth=2)
plt.title('Model Accuracy Optimization Metrics', fontsize=14, fontweight='bold')
plt.xlabel('Epochs Run', fontsize=11)
plt.ylabel('Accuracy Percentage (0.0 - 1.0)', fontsize=11)
plt.grid(True, linestyle=':', alpha=0.6)
plt.legend(fontsize=11)

plt.tight_layout()
```
